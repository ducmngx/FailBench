# LIBERO Failure Gallery (v2)

Visual confirmation that failure modes are causing the *expected* breakdown. v2 changes from v1:

1. **Joint coverage**: instead of clustering on j4, the gallery covers the three pitch joints (j2 shoulder, j4 elbow, j6 wrist) and two rotation joints (j1, j7) so the per-joint failure signatures are distinguishable.
2. **Active resistance A/B**: each pitch single, each multi, ALL_JOINTS, and the gripper failures are rendered under both `none` (current behavior — torque-controlled arm sags passively) and `gravcomp_pd` (healthy joints actively hold pose via gravity-comp + PD). The pair makes the difference legible.

Generate more videos via `scripts/libero/record_failure_video.py --resistance {none,gravcomp_pd}`.

```python
import os, sys, glob
from IPython.display import Video, display, HTML

REPO_ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))
os.chdir(REPO_ROOT)
sys.path.insert(0, REPO_ROOT)

VIDEO_DIR = "out/libero_videos"
videos = sorted(glob.glob(os.path.join(VIDEO_DIR, "*.mp4")))
print(f"{len(videos)} videos in {VIDEO_DIR}")
```

```python
def show_clip(path, width=380):
    return Video(path, embed=True, width=width,
                 html_attributes="controls loop muted autoplay")

def pair(name_none, name_pd, width=380):
    """Side-by-side A/B for the same trial under each resistance mode."""
    p_none = os.path.join(VIDEO_DIR, name_none)
    p_pd = os.path.join(VIDEO_DIR, name_pd)
    cells = []
    for label, path in [("resistance=none", p_none), ("resistance=gravcomp_pd", p_pd)]:
        if not os.path.exists(path):
            cells.append(f'<div style="display:inline-block;margin:6px;color:#a00">missing: {path}</div>')
            continue
        v = show_clip(path, width=width)
        cells.append(
            f'<div style="display:inline-block;margin:6px;text-align:center;vertical-align:top">'
            f'<div style="font-family:monospace;font-size:12px">{label}<br>{os.path.basename(path)}</div>'
            f'{v._repr_html_()}</div>'
        )
    display(HTML("".join(cells)))

def single(name, width=420):
    p = os.path.join(VIDEO_DIR, name)
    if not os.path.exists(p):
        display(HTML(f'<div style="color:#a00">missing: {p}</div>'))
        return
    display(HTML(
        f'<div style="text-align:center">'
        f'<div style="font-family:monospace;font-size:12px">{name}</div>'
        f'{show_clip(p, width=width)._repr_html_()}</div>'
    ))
```

## Pitch joints (gravity-loaded)

These are the high-signal failures: each one drops the arm under gravity in a *distinctive* way. Under `none` the rest of the arm goes limp too, masking the signature; under `gravcomp_pd` only the named joint droops while the other six hold rigid.

- **j2 (shoulder pitch)** — heaviest gravity load. Whole arm collapses forward.
- **j4 (elbow pitch)** — moderate. Forearm + EE swing down.
- **j6 (wrist pitch)** — light. EE pitches down, arm body holds.

```python
display(HTML("<h4>SINGLE_JOINT(joint2) — shoulder pitch</h4>"))
pair("pitch_joint2_none.mp4", "pitch_joint2_gravcomp_pd.mp4")
```

```python
display(HTML("<h4>SINGLE_JOINT(joint4) — elbow pitch</h4>"))
pair("pitch_joint4_none.mp4", "pitch_joint4_gravcomp_pd.mp4")
```

```python
display(HTML("<h4>SINGLE_JOINT(joint6) — wrist pitch</h4>"))
pair("pitch_joint6_none.mp4", "pitch_joint6_gravcomp_pd.mp4")
```

## Rotation joints (no gravity load)

These failures are deliberately subtle — kept in the dataset at low weight (3 % j1, 2 % j7) for diversity, not because they're dramatic. Shown under `gravcomp_pd` only so that the (small) joint-specific motion isn't swamped by global arm collapse.

```python
display(HTML("<h4>SINGLE_JOINT(joint1) — base yaw, vertical axis</h4>"))
single("rotation_joint1_gravcomp_pd.mp4")
```

```python
display(HTML("<h4>SINGLE_JOINT(joint7) — wrist twist, EE roll</h4>"))
single("rotation_joint7_gravcomp_pd.mp4")
```

## ALL_JOINTS — full system death

Every arm joint is in `failed_joint_ids`, so the resistance loop has *no* healthy joint to drive — the two modes are equivalent here. Use this as the visual baseline for what "completely limp" looks like.

## Gripper failures — the most striking A/B

**Question from review:** "Why does GRIPPER_OPEN look like ALL_JOINTS?"

Answer: under `none`, the gripper opens *and* the arm sags from zero torque on the arm actuators — the silhouette matches a full collapse. Under `gravcomp_pd`, the arm holds rigid and only the bowl drops out of the opening fingers, which is the correct visual for that failure mode.

If you're going to compare one pair from this notebook, make it this one.

```python
display(HTML("<h4>GRIPPER_OPEN — fingers fully open</h4>"))
pair("gripper_gripper_open_none.mp4", "gripper_gripper_open_gravcomp_pd.mp4")
```
