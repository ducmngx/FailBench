```python
from planner.utils.traj_saver import ExperimentTrajectoryManager
import os
import pandas as pd
```

```python

trajectory_folders = ["scene1_trajs", "scene2_trajs", "scene3_trajs"]
repo_path = os.getcwd()

raw_dict = {"scene":[], "sample":[], "traj_len":[], "me_cost":[], "safety_cost":[]}


exp_traj_manager = ExperimentTrajectoryManager()
for folder in trajectory_folders:
    traj_folder = os.path.join(repo_path, "collected_trajs", folder)
    traj_files = [os.path.join(traj_folder, f) for f in os.listdir(os.path.join(traj_folder))]
    traj_files.sort()
    for file in traj_files:
        fname = os.path.splitext(os.path.basename(file))[0]
        sample_no = fname.split("_")[-1]
        print(fname, sample_no)
        exp_traj_manager.load_from_file(file)

        for scene, scene_dict in exp_traj_manager.trajectories.items():

            traj_len = 0    
            me_cost = 0
            safety_cost = 0
            for phase, phase_dict in scene_dict.items():
                traj_len += len(phase_dict["trajectory"]) -1
                me_cost += phase_dict["me_cost"][1:].sum()
                safety_cost += phase_dict["safety_cost"][1:].sum()
            
            print(f"traj_len {traj_len} me_cost {me_cost} safety_cost {safety_cost}")
            raw_dict["scene"].append(scene)
            raw_dict["sample"].append(sample_no)
            raw_dict["traj_len"].append(traj_len)
            raw_dict["me_cost"].append(me_cost)
            raw_dict["safety_cost"].append(safety_cost)

        print("------------")
```

```python
exp_traj_manager.report()
```

```python
df = pd.DataFrame(raw_dict)
df["total_cost"] = df["me_cost"] + df["safety_cost"]

df["avg_me_cost"] = df["me_cost"] / df["traj_len"]
df["avg_safety_cost"] = df["safety_cost"] / df["traj_len"]
df["avg_total_cost"] = df["total_cost"] / df["traj_len"]

df.sort_values(by=["scene", "total_cost"], ascending=[True, True], inplace=True)
df
```

```python
df.sort_values(by=["scene", "avg_total_cost"], ascending=[True, True], inplace=True)
df
```
